```python
import os
import json
import pandas as pd
import psycopg2
```

```python
path5="C:/Users/saman/Desktop/PhonePe/phonepe-project/pulse/data/aggregated/insurance/country/india/state/"
agg_ins_list=os.listdir(path5)
columns5={"States":[],"Years":[],"Quarter":[],"Transaction_type":[],"Transaction_count":[],"Transaction_amount":[]}
for state in agg_ins_list:
    curr_states=path5+state+"/"
    agg_year_list=os.listdir(curr_states)
    for year in agg_year_list:
        curr_year=curr_states+year+"/"
        agg_file_list=os.listdir(curr_year)
        for file in agg_file_list:
            curr_file=curr_year+file
            data=open(curr_file,"r")
            E=json.load(data)
            for i in E["data"]["transactionData"]:
                name=i["name"]
                count=i["paymentInstruments"][0]["count"]
                amount=i["paymentInstruments"][0]["amount"]
                columns5["Transaction_type"].append(name)
                columns5["Transaction_count"].append(count)
                columns5["Transaction_amount"].append(amount)
                columns5["States"].append(state)
                columns5["Years"].append(year)
                columns5["Quarter"].append(int(file.strip(".json")))
```

```python
aggr_insurance=pd.DataFrame(columns5)
aggr_insurance["States"]=aggr_insurance["States"].str.replace("andaman-&-nicobar-islands","Andaman & Nicobar")
aggr_insurance["States"]=aggr_insurance["States"].str.replace("-"," ")
aggr_insurance["States"]=aggr_insurance['States'].str.title()
aggr_insurance["States"]=aggr_insurance["States"].str.replace("Dadra & Nagar Haveli & Daman & Diu","Dadra and Nagar Haveli and Daman and Diu")
```
